```
#2-a Convert the last column strings to integers, as
# setosa -> 0
# versicolor -> 1
# virginica -> 2
```

```
iris.iloc[:, -1] = iris.iloc[:, -1].map({
    "setosa": 0,
    "versicolor": 1,
    "virginica": 2
})
```

```
#03-a Select the first four columns of the dataframe as the Data and call the data x.
```

```
x = iris.iloc[:, 0:4]
x.head()
```

```
#03-b Select the last column as the label and call it y.
```

```
y = iris.iloc[:, -1]
y.head()
```

```
#04 Now, it is the time to split the data (x) and the label (y) into training and testin segments,
#   and call them parts x_train,x_test,y_train,y_test.
```

```
#04-a Split the data, and label ito training and testing segments, while the size of test data is 20%,
#     and use randome_state = 123 to make all the answers similar in the whole class.
```

```
from sklearn.model_selection import train_test_split

x_train, x_test, y_train, y_test = train_test_split(
    x, y,
    test_size=0.2,
    random_state=123
)
```

```
#05-a Convert y_train and y_test from categorical to one-hot encoded values, and call them
#     one_hot_train_labels, and one_hot_test_labels, respectively.
```

```
from tensorflow.keras.utils import to_categorical

one_hot_train_labels = to_categorical(y_train)
one_hot_test_labels = to_categorical(y_test)
one_hot_train_labels[:5]
```

```
#06-a Define a sequential model that contains 3 dense layers.
#     In the first and second layers, use 100 hidden units (neurons) and a 'relu' activation function.
#     For the last layer, we need to use 3 as the number of output,
#     and the reason is we have three classes (setosa -> 0, versicolor -> 1, virginica -> 2)
#     Also, please pick an appropriate loss function for the last layer.
#     (You may already have guessed what could be the best loss function when we have converted the categorical values to encoded values.)
```

```
from tensorflow.keras import Sequential
from tensorflow.keras.layers import Dense, Input

model = Sequential([
    Input(shape=(x_train.shape[1],)),
    Dense(100, activation='relu'),
    Dense(100, activation='relu'),
    Dense(3, activation='softmax')
])
```

```
#06-b Now, compile your model using the 'rmsprop' optimizer, an appropriate loss function, and 'accuracy' for your metrics.
```

```
model.compile(
    optimizer='rmsprop',
    loss='categorical_crossentropy',
    metrics=['accuracy']
)
```

```
#07-a Train your model using epoch=20, batch=10.
```

```
history = model.fit(
    x_train,
    one_hot_train_labels,
    epochs=20,
    batch_size=10
)
```

```
#08-a Calculate the result, evaluating x_test and one_hot_test_label, and report on the accuracy of your model.
```

```
test_loss, test_accuracy = model.evaluate(x_test, one_hot_test_labels)
print("Test Loss:", test_loss)
print("Test Accuracy:", test_accuracy)
```

```
#09-a Now, we repeat all we have done in #06-a with the following parameters:
#     In the first and second hidden layers, use 10 hidden units (neurons) and a 'relu' activation function.
#     For the output layer, we need to use 3.
```

```
from tensorflow.keras import Sequential
from tensorflow.keras.layers import Dense, Input
model_10 = Sequential([
    Input(shape=(x_train.shape[1],)),
    Dense(10, activation='relu'),
    Dense(10, activation='relu'),
    Dense(3, activation='softmax')
])
model_10.compile(
    optimizer='rmsprop',
    loss='categorical_crossentropy',
    metrics=['accuracy']
)
```

```
#09-b Use the same settings as for #06-b (compile using the same optimizer, loss function) and output the accuracy
```

```
history_10 = model_10.fit(
    x_train,
    one_hot_train_labels,
    epochs=20,
    batch_size=10
)
test_loss_10, test_accuracy_10 = model_10.evaluate(x_test, one_hot_test_labels)
print("Test Accuracy (10 neurons, 20 epochs):", test_accuracy_10)
```

```
#09-c Please fit your model using epoch=4, batch=10.
```

```
test_loss_10_short, test_accuracy_10_short = model_10.evaluate(
    x_test,
    one_hot_test_labels
)

print("Test Accuracy (10 neurons, 4 epochs):", test_accuracy_10_short)
```

ANSWER:
Increasing the number of layers or neurons increases model capacity, which can improve accuracy, but if the model becomes too complex it can overfit. 
We detect overfitting when training accuracy is much higher than test accuracy, and we reduce it by simplifying the model, training for fewer epochs, or using certain techniques like dropout.

```
#10-b
# Use different settings than the ones we chose in #09
# Change each of the following once:
#   i)   optimizer
#   ii)  number of layers
#   iii) hidden units
#   iv)  number of epochs
#   v)   batch_size
# Change one parameter at a time, and rerun the cell(s), then explain the output
```

```
#Baseline reference
from tensorflow.keras import Sequential
from tensorflow.keras.layers import Dense, Input

def build_model(hidden_units=(10,10), num_hidden_layers=2):
    model = Sequential()
    model.add(Input(shape=(x_train.shape[1],)))
    # hidden layers
    for i in range(num_hidden_layers):
        model.add(Dense(hidden_units[i], activation='relu'))
    # output layer
    model.add(Dense(3, activation='softmax'))
    return model

# Baseline settings from #09
baseline_model = build_model(hidden_units=(10,10), num_hidden_layers=2)
baseline_model.compile(optimizer='rmsprop', loss='categorical_crossentropy', metrics=['accuracy'])
baseline_model.fit(x_train, one_hot_train_labels, epochs=20, batch_size=10, verbose=0)

baseline_loss, baseline_acc = baseline_model.evaluate(x_test, one_hot_test_labels, verbose=0)
print("BASELINE (rmsprop, 2 layers, 10 units, epochs=20, batch=10) accuracy:", baseline_acc)
```

```
#10-b-i) Change optimizer
model_opt = build_model(hidden_units=(10,10), num_hidden_layers=2)
model_opt.compile(optimizer='adam', loss='categorical_crossentropy', metrics=['accuracy'])
model_opt.fit(x_train, one_hot_train_labels, epochs=20, batch_size=10, verbose=0)

loss_opt, acc_opt = model_opt.evaluate(x_test, one_hot_test_labels, verbose=0)
print("OPTIMIZER CHANGED (adam) accuracy:", acc_opt)
```

```
#10-b-ii) Change number of layers
#Added 1 more hidden layer, 3 total
model_layers = build_model(hidden_units=(10,10,10), num_hidden_layers=3)
model_layers.compile(optimizer='rmsprop', loss='categorical_crossentropy', metrics=['accuracy'])
model_layers.fit(x_train, one_hot_train_labels, epochs=20, batch_size=10, verbose=0)

loss_layers, acc_layers = model_layers.evaluate(x_test, one_hot_test_labels, verbose=0)
print("LAYERS CHANGED (3 hidden layers) accuracy:", acc_layers)
```

```
#10-b-iii) Change hidden units
#Change 10 -> 50 in both layer
model_units = build_model(hidden_units=(50,50), num_hidden_layers=2)
model_units.compile(optimizer='rmsprop', loss='categorical_crossentropy', metrics=['accuracy'])
model_units.fit(x_train, one_hot_train_labels, epochs=20, batch_size=10, verbose=0)

loss_units, acc_units = model_units.evaluate(x_test, one_hot_test_labels, verbose=0)
print("HIDDEN UNITS CHANGED (50,50) accuracy:", acc_units)
```

```
#10-b-iv) Change number of epochs
#Change 20 -> 40 epochs
model_epochs = build_model(hidden_units=(10,10), num_hidden_layers=2)
model_epochs.compile(optimizer='rmsprop', loss='categorical_crossentropy', metrics=['accuracy'])
model_epochs.fit(x_train, one_hot_train_labels, epochs=40, batch_size=10, verbose=0)

loss_epochs, acc_epochs = model_epochs.evaluate(x_test, one_hot_test_labels, verbose=0)
print("EPOCHS CHANGED (40) accuracy:", acc_epochs)
```

```
#10-b-v) Change batch_size
#Change batch_size 10 -> 20
model_batch = build_model(hidden_units=(10,10), num_hidden_layers=2)
model_batch.compile(optimizer='rmsprop', loss='categorical_crossentropy', metrics=['accuracy'])
model_batch.fit(x_train, one_hot_train_labels, epochs=20, batch_size=20, verbose=0)

loss_batch, acc_batch = model_batch.evaluate(x_test, one_hot_test_labels, verbose=0)
print("BATCH SIZE CHANGED (20) accuracy:", acc_batch)
```
