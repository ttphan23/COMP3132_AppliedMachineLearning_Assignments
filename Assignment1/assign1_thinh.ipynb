```python
# (Load dataset)
from keras.datasets import fashion_mnist
import numpy as np
import matplotlib.pyplot as plt

(train_images, train_labels), (test_images, test_labels) = fashion_mnist.load_data()

class_names = [
    "T-shirt/top", "Trouser", "Pullover", "Dress", "Coat",
    "Sandal", "Shirt", "Sneaker", "Bag", "Ankle boot"
]
```

```python
# Question 1: dimensions (shapes)
print("train_images shape:", train_images.shape)
print("train_labels shape:", train_labels.shape)
print("test_images shape:", test_images.shape)
print("test_labels shape:", test_labels.shape)
```

```python
# Question 2: lengths
print("len(train_labels):", len(train_labels))
print("len(test_labels):", len(test_labels))
```

```python
# Question 10: import models and layers
from keras import models, layers
```

```python
# Question 11-16: define Sequential model (myNetwork), flatten, normalize, 2 hidden layers, output softmax
# Normalize (Question 13)
train_images_norm = train_images.astype("float32") / 255.0
test_images_norm = test_images.astype("float32") / 255.0
```

```python
# Build model (Questions 11, 12, 14, 15, 16)
myNetwork = models.Sequential([
    layers.Flatten(input_shape=(28, 28)),        # Q12 flatten 28x28 -> 784
    layers.Dense(512, activation="relu"),        # Q14
    layers.Dense(128, activation="relu"),        # Q15
    layers.Dense(10, activation="softmax")       # Q16
])
```

```python
# Question 17-18 Option A (adam + sparse_categorical_crossentropy)
myNetwork_A = models.Sequential([
    layers.Flatten(input_shape=(28, 28)),
    layers.Dense(512, activation="relu"),
    layers.Dense(128, activation="relu"),
    layers.Dense(10, activation="softmax")
])

myNetwork_A.compile(
    optimizer="adam",
    loss="sparse_categorical_crossentropy",
    metrics=["accuracy"]
)

history_A = myNetwork_A.fit(
    train_images_norm, train_labels,
    epochs=10,
    validation_split=0.2,
    verbose=1
)
```

```python
# Question 17-18 Option B (rmsprop + categorical_crossentropy)
# (Needs one-hot labels)
myNetwork_B = models.Sequential([
    layers.Flatten(input_shape=(28, 28)),
    layers.Dense(512, activation="relu"),
    layers.Dense(128, activation="relu"),
    layers.Dense(10, activation="softmax")
])

# One-hot encode labels for categorical_crossentropy
train_labels_oh = np.eye(10)[train_labels]
test_labels_oh = np.eye(10)[test_labels]

myNetwork_B.compile(
    optimizer="rmsprop",
    loss="categorical_crossentropy",
    metrics=["accuracy"]
)

history_B = myNetwork_B.fit(
    train_images_norm, train_labels_oh,
    epochs=10,
    validation_split=0.2,
    verbose=1
)
```

Question 19 – Fashion MNIST vs MNIST

Fashion MNIST is harder than MNIST because clothing items look more similar to each other than handwritten digits. Digits have clear shapes, while clothes share similar patterns and textures. Because of this, models usually get lower accuracy on Fashion MNIST.

Question 20 – Accuracy differences

Fashion MNIST has lower accuracy because its classes are more visually similar, making them harder to distinguish. Clothing items share shapes and details, which makes feature learning more difficult. This is why the model performs better on MNIST than Fashion MNIST.

```python
# Question 21 (evaluate) for Option A
test_loss_A, test_acc_A = myNetwork_A.evaluate(test_images_norm, test_labels, verbose=0)
print("\nOption A - Test loss:", test_loss_A)
print("Option A - Test accuracy:", test_acc_A)
```

```python
# Question 21 (evaluate) for Option B
test_loss_B, test_acc_B = myNetwork_B.evaluate(test_images_norm, test_labels_oh, verbose=0)
print("\nOption B - Test loss:", test_loss_B)
print("Option B - Test accuracy:", test_acc_B)
```
